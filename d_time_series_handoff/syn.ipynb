```python


import os
import re
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt
import seaborn as sns

from tqdm import tqdm
from transformers import AutoTokenizer, AutoModel
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.metrics.pairwise import cosine_similarity


# =========================
# 1. 基本設定
# =========================
# from google.colab import drive
# drive.mount('/content/drive')

FILE_PATH = "D:\\時間序列\\DECEN\\post_level_dataset_800_plus_syn.csv"

SAVE_DIR = "D:\\時間序列\\DECEN\\quality_analysis"
os.makedirs(SAVE_DIR, exist_ok=True)

MODEL_NAME = "bert-base-chinese"
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print("DEVICE:", DEVICE)


# =========================
# 2. 讀取資料
# =========================
df = pd.read_csv(FILE_PATH)

print("資料筆數:", len(df))
print("欄位:", df.columns.tolist())
display(df.head())


# =========================
# 3. 自動偵測文字欄位
# 若偵測錯，請手動改 TEXT_COL
# =========================
possible_text_cols = ["caption", "text", "content", "post", "main_text", "sentence"]

TEXT_COL = None
for col in possible_text_cols:
    if col in df.columns:
        TEXT_COL = col
        break

if TEXT_COL is None:
    raise ValueError("找不到文字欄位，請手動設定 TEXT_COL，例如 TEXT_COL = 'caption'")

print("使用文字欄位:", TEXT_COL)

df[TEXT_COL] = df[TEXT_COL].fillna("").astype(str)

# 前 800 真實，後 200 合成
# df["data_type"] = ["real"] * 800 + ["synthetic"] * (len(df) - 800)
# =========================
# 依照「後 200 筆為合成，其餘前面為真實」切分
# =========================
N_SYN = 200

df = df.reset_index(drop=True)

if len(df) <= N_SYN:
    raise ValueError(f"資料總筆數只有 {len(df)}，不足以切出後 {N_SYN} 筆合成資料")

df["data_type"] = "real"
df.loc[df.index[-N_SYN:], "data_type"] = "synthetic"

print(df["data_type"].value_counts())
print("真實資料範圍 index:", df[df["data_type"] == "real"].index.min(), "~", df[df["data_type"] == "real"].index.max())
print("合成資料範圍 index:", df[df["data_type"] == "synthetic"].index.min(), "~", df[df["data_type"] == "synthetic"].index.max())



# =========================
# 4. 簡單文字清理
# =========================
def clean_text(text):
    text = str(text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

df["clean_text"] = df[TEXT_COL].apply(clean_text)

df["char_len"] = df["clean_text"].str.len()

display(df.groupby("data_type")["char_len"].describe())


# =========================
# 5. 載入 BERT / DER encoder 基底
# 這裡用 bert-base-chinese 做 embedding 分析
# =========================
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModel.from_pretrained(MODEL_NAME).to(DEVICE)
model.eval()


def get_bert_embeddings(texts, batch_size=32, max_length=128):
    all_embeddings = []

    for i in tqdm(range(0, len(texts), batch_size)):
        batch_texts = texts[i:i+batch_size]

        encoded = tokenizer(
            batch_texts,
            padding=True,
            truncation=True,
            max_length=max_length,
            return_tensors="pt"
        ).to(DEVICE)

        with torch.no_grad():
            outputs = model(**encoded)

        # mean pooling
        last_hidden = outputs.last_hidden_state
        attention_mask = encoded["attention_mask"].unsqueeze(-1)

        masked_hidden = last_hidden * attention_mask
        sum_hidden = masked_hidden.sum(dim=1)
        lengths = attention_mask.sum(dim=1)

        embeddings = sum_hidden / lengths
        all_embeddings.append(embeddings.cpu().numpy())

    return np.vstack(all_embeddings)


texts = df["clean_text"].tolist()
embeddings = get_bert_embeddings(texts)

print("embedding shape:", embeddings.shape)

np.save(os.path.join(SAVE_DIR, "bert_embeddings.npy"), embeddings)
df.to_csv(os.path.join(SAVE_DIR, "dataset_with_type.csv"), index=False, encoding="utf-8-sig")


# =========================
# 6. Embedding 分布分析：PCA
# =========================
pca = PCA(n_components=2, random_state=42)
pca_2d = pca.fit_transform(embeddings)

df["pca_x"] = pca_2d[:, 0]
df["pca_y"] = pca_2d[:, 1]

plt.figure(figsize=(8, 6))

for dtype in ["real", "synthetic"]:
    subset = df[df["data_type"] == dtype]
    plt.scatter(
        subset["pca_x"],
        subset["pca_y"],
        label=dtype,
        alpha=0.65,
        s=25
    )

plt.title("PCA Embedding Distribution: Real vs Synthetic")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend()
plt.tight_layout()

pca_path = os.path.join(SAVE_DIR, "embedding_distribution_pca.png")
plt.savefig(pca_path, dpi=300)
plt.show()

print("PCA 圖已儲存：", pca_path)


# # =========================
# # 7. Embedding 分布分析：t-SNE
# # =========================
# tsne = TSNE(
#     n_components=2,
#     perplexity=30,
#     learning_rate="auto",
#     init="pca",
#     random_state=42
# )

# tsne_2d = tsne.fit_transform(embeddings)

# df["tsne_x"] = tsne_2d[:, 0]
# df["tsne_y"] = tsne_2d[:, 1]

# plt.figure(figsize=(8, 6))

# for dtype in ["real", "synthetic"]:
#     subset = df[df["data_type"] == dtype]
#     plt.scatter(
#         subset["tsne_x"],
#         subset["tsne_y"],
#         label=dtype,
#         alpha=0.65,
#         s=25
#     )

# plt.title("t-SNE Embedding Distribution: Real vs Synthetic")
# plt.xlabel("t-SNE 1")
# plt.ylabel("t-SNE 2")
# plt.legend()
# plt.tight_layout()

# tsne_path = os.path.join(SAVE_DIR, "embedding_distribution_tsne.png")
# plt.savefig(tsne_path, dpi=300)
# plt.show()

# print("t-SNE 圖已儲存：", tsne_path)

# =========================
# 7. Embedding 分布分析：改用 UMAP
# =========================
!pip install -q umap-learn

import umap

umap_model = umap.UMAP(
    n_components=2,
    n_neighbors=15,
    min_dist=0.1,
    metric="cosine",
    random_state=42
)

umap_2d = umap_model.fit_transform(embeddings)

df["umap_x"] = umap_2d[:, 0]
df["umap_y"] = umap_2d[:, 1]

plt.figure(figsize=(8, 6))

for dtype in ["real", "synthetic"]:
    subset = df[df["data_type"] == dtype]
    plt.scatter(
        subset["umap_x"],
        subset["umap_y"],
        label=dtype,
        alpha=0.65,
        s=25
    )

plt.title("UMAP Embedding Distribution: Real vs Synthetic")
plt.xlabel("UMAP 1")
plt.ylabel("UMAP 2")
plt.legend()
plt.tight_layout()

umap_path = os.path.join(SAVE_DIR, "embedding_distribution_umap.png")
plt.savefig(umap_path, dpi=300)
plt.show()

print("UMAP 圖已儲存：", umap_path)
# =========================
# 8. 去重率分析：完全重複
# =========================
real_df = df[df["data_type"] == "real"].copy()
syn_df = df[df["data_type"] == "synthetic"].copy()

exact_duplicate_all = df["clean_text"].duplicated().sum()
exact_duplicate_syn_internal = syn_df["clean_text"].duplicated().sum()
exact_duplicate_syn_vs_real = syn_df["clean_text"].isin(real_df["clean_text"]).sum()

exact_summary = {
    "total_rows": len(df),
    "real_rows": len(real_df),
    "synthetic_rows": len(syn_df),
    "exact_duplicate_all_count": int(exact_duplicate_all),
    "exact_duplicate_all_rate": exact_duplicate_all / len(df),
    "synthetic_internal_exact_duplicate_count": int(exact_duplicate_syn_internal),
    "synthetic_internal_exact_duplicate_rate": exact_duplicate_syn_internal / len(syn_df),
    "synthetic_vs_real_exact_duplicate_count": int(exact_duplicate_syn_vs_real),
    "synthetic_vs_real_exact_duplicate_rate": exact_duplicate_syn_vs_real / len(syn_df)
}

exact_summary_df = pd.DataFrame([exact_summary])
display(exact_summary_df)

exact_summary_df.to_csv(
    os.path.join(SAVE_DIR, "exact_duplicate_summary.csv"),
    index=False,
    encoding="utf-8-sig"
)


# =========================
# 9. 去重率分析：語意近似重複
# threshold 可依需求調整
# 0.90 = 較寬鬆
# 0.95 = 較嚴格
# =========================
real_embeddings = embeddings[df["data_type"].values == "real"]
syn_embeddings = embeddings[df["data_type"].values == "synthetic"]

sim_syn_real = cosine_similarity(syn_embeddings, real_embeddings)

thresholds = [0.90, 0.95]

semantic_results = []

for th in thresholds:
    max_sim = sim_syn_real.max(axis=1)
    duplicate_flag = max_sim >= th

    semantic_results.append({
        "threshold": th,
        "synthetic_vs_real_semantic_duplicate_count": int(duplicate_flag.sum()),
        "synthetic_vs_real_semantic_duplicate_rate": duplicate_flag.mean(),
        "mean_max_similarity_to_real": float(max_sim.mean()),
        "median_max_similarity_to_real": float(np.median(max_sim)),
        "max_similarity_to_real": float(max_sim.max())
    })

semantic_summary_df = pd.DataFrame(semantic_results)
display(semantic_summary_df)

semantic_summary_df.to_csv(
    os.path.join(SAVE_DIR, "semantic_duplicate_summary.csv"),
    index=False,
    encoding="utf-8-sig"
)


# =========================
# 10. 找出合成資料最像哪一筆真實資料
# =========================
max_real_idx = sim_syn_real.argmax(axis=1)
max_real_sim = sim_syn_real.max(axis=1)

similar_pairs = syn_df.copy().reset_index().rename(columns={"index": "synthetic_original_index"})
real_reset = real_df.copy().reset_index().rename(columns={"index": "real_original_index"})

similar_pairs["most_similar_real_index"] = real_reset.iloc[max_real_idx]["real_original_index"].values
similar_pairs["max_similarity_to_real"] = max_real_sim
similar_pairs["most_similar_real_text"] = real_reset.iloc[max_real_idx]["clean_text"].values

similar_pairs = similar_pairs[
    [
        "synthetic_original_index",
        "clean_text",
        "most_similar_real_index",
        "most_similar_real_text",
        "max_similarity_to_real"
    ]
].rename(columns={
    "clean_text": "synthetic_text"
})

similar_pairs = similar_pairs.sort_values(
    "max_similarity_to_real",
    ascending=False
)

display(similar_pairs.head(20))

# similar_pairs.to_csv(
#     os.path.join(SAVE_DIR, "synthetic_most_similar_real_pairs.csv"),
#     index=False,
#     encoding="utf-8-sig"
# )


# =========================
# 11. 合成資料內部語意重複率
# =========================
sim_syn_syn = cosine_similarity(syn_embeddings, syn_embeddings)

# 排除自己跟自己的相似度
np.fill_diagonal(sim_syn_syn, -1)

internal_results = []

for th in thresholds:
    max_internal_sim = sim_syn_syn.max(axis=1)
    internal_duplicate_flag = max_internal_sim >= th

    internal_results.append({
        "threshold": th,
        "synthetic_internal_semantic_duplicate_count": int(internal_duplicate_flag.sum()),
        "synthetic_internal_semantic_duplicate_rate": internal_duplicate_flag.mean(),
        "mean_internal_max_similarity": float(max_internal_sim.mean()),
        "median_internal_max_similarity": float(np.median(max_internal_sim)),
        "max_internal_similarity": float(max_internal_sim.max())
    })

internal_summary_df = pd.DataFrame(internal_results)
display(internal_summary_df)

internal_summary_df.to_csv(
    os.path.join(SAVE_DIR, "synthetic_internal_semantic_duplicate_summary.csv"),
    index=False,
    encoding="utf-8-sig"
)


# =========================
# 12. 人工品質評分表
# =========================
# 評分欄位：
# fluency_score: 流暢度 1-5
# realism_score: 真實貼文感 1-5
# label_consistency_score: 是否符合指定情緒類別 1-5
# diversity_score: 多樣性 1-5
# usable_score: 是否適合放入訓練資料 1-5
# note: 備註

human_eval_df = syn_df.copy().reset_index().rename(columns={"index": "original_index"})

human_eval_df = human_eval_df[["original_index", "clean_text"]].rename(
    columns={"clean_text": "synthetic_text"}
)

human_eval_df["fluency_score"] = ""
human_eval_df["realism_score"] = ""
human_eval_df["label_consistency_score"] = ""
human_eval_df["diversity_score"] = ""
human_eval_df["usable_score"] = ""
human_eval_df["note"] = ""

human_eval_path = os.path.join(SAVE_DIR, "human_quality_rating_template.csv")

human_eval_df.to_csv(
    human_eval_path,
    index=False,
    encoding="utf-8-sig"
)

print("人工品質評分表已輸出：", human_eval_path)
display(human_eval_df.head())


# =========================
# 13. 若人工評分表已填完，可跑這段統計
# =========================
# 修改成你填完後的檔案路徑
RATED_PATH = human_eval_path

rated_df = pd.read_csv(RATED_PATH)

score_cols = [
    "fluency_score",
    "realism_score",
    "label_consistency_score",
    "diversity_score",
    "usable_score"
]

has_scores = all(
    pd.api.types.is_numeric_dtype(pd.to_numeric(rated_df[col], errors="coerce"))
    for col in score_cols
)

for col in score_cols:
    rated_df[col] = pd.to_numeric(rated_df[col], errors="coerce")

if rated_df[score_cols].notna().sum().sum() > 0:
    quality_summary = rated_df[score_cols].describe().T
    quality_summary["mean"] = rated_df[score_cols].mean()
    quality_summary["std"] = rated_df[score_cols].std()

    display(quality_summary)

    quality_summary.to_csv(
        os.path.join(SAVE_DIR, "human_quality_rating_summary.csv"),
        encoding="utf-8-sig"
    )

    rated_df["overall_quality_score"] = rated_df[score_cols].mean(axis=1)

    plt.figure(figsize=(7, 5))
    plt.hist(rated_df["overall_quality_score"].dropna(), bins=10)
    plt.title("Synthetic Data Human Quality Score Distribution")
    plt.xlabel("Overall Quality Score")
    plt.ylabel("Count")
    plt.tight_layout()

    quality_fig_path = os.path.join(SAVE_DIR, "human_quality_score_distribution.png")
    plt.savefig(quality_fig_path, dpi=300)
    plt.show()

    print("人工品質評分統計已完成")
else:
    print("目前尚未填入人工評分，因此只輸出評分表。")


# =========================
# 14. 匯出總結果
# =========================
df.to_csv(
    os.path.join(SAVE_DIR, "embedding_distribution_coordinates.csv"),
    index=False,
    encoding="utf-8-sig"
)

print("全部分析完成，結果儲存在：", SAVE_DIR)
```

```python
import re

def clean_for_human_eval(text):

    text = str(text)

    # 移除網址
    text = re.sub(r"http\S+|www\S+", "", text)

    # 移除 hashtag
    text = re.sub(r"#\S+", "", text)

    # 移除 @mention
    text = re.sub(r"@\S+", "", text)

    # 換行轉空白
    text = text.replace("\n", " ")

    # 移除多餘空白
    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

#### 一致性分析

```python
# 兩份評分檔
# =========================
# 0. 匯入套件
# =========================
import os
import pandas as pd
import numpy as np
from sklearn.metrics import cohen_kappa_score, accuracy_score, confusion_matrix, classification_report
from scipy.stats import spearmanr


# =========================
# 1. 檔案路徑
# =========================
RATER1_PATH = r"D:\時間序列\DECEN\human_quality_rating\human_quality_rating_blind_test_n.xlsx"
RATER2_PATH = r"D:\時間序列\DECEN\human_quality_rating\human_quality_rating_blind_test_w.xlsx"

SAVE_DIR = r"D:\時間序列\DECEN\human_quality_rating\results"
os.makedirs(SAVE_DIR, exist_ok=True)


# =========================
# 2. 讀取資料
# =========================
r1 = pd.read_excel(RATER1_PATH)
r2 = pd.read_excel(RATER2_PATH)

print("Rater1 欄位:", r1.columns.tolist())
print("Rater2 欄位:", r2.columns.tolist())

# display(r1.head())
# display(r2.head())


# =========================
# 3. 欄位設定
# =========================
ID_COL = "sample_id"

score_cols = [
    "fluency_score",
    "realism_score",
    "label_consistency_score"
]

PRED_COL = "predicted_source"
SOURCE_COL = "true_source"


# =========================
# 4. 檢查必要欄位
# =========================
for col in [ID_COL, SOURCE_COL, PRED_COL] + score_cols:
    if col not in r1.columns:
        raise ValueError(f"Rater1 缺少欄位：{col}")

for col in [ID_COL, PRED_COL] + score_cols:
    if col not in r2.columns:
        raise ValueError(f"Rater2 缺少欄位：{col}")


# =========================
# 5. 用 sample_id 將 rater2 補上 true_source
# =========================
r2 = r2.merge(
    r1[[ID_COL, SOURCE_COL]],
    on=ID_COL,
    how="left"
)

print("Rater2 補回 true_source 後：")
print(r2[SOURCE_COL].value_counts(dropna=False))


# =========================
# 6. 轉換欄位格式
# =========================
def normalize_source(x):
    if pd.isna(x):
        return np.nan

    x = str(x).strip().lower()

    if x in ["0", "real", "真實"]:
        return 0
    elif x in ["1", "synthetic", "合成"]:
        return 1
    else:
        return np.nan


for df in [r1, r2]:
    for col in score_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    df["overall_quality_score"] = df[score_cols].mean(axis=1)
    df["true_source_code"] = df[SOURCE_COL].apply(normalize_source)
    df["predicted_source_code"] = df[PRED_COL].apply(normalize_source)


# =========================
# 7. 各評分者品質摘要
# =========================
def quality_summary(df, name):
    summary = df[score_cols + ["overall_quality_score"]].agg(
        ["mean", "std", "min", "max"]
    ).T.round(3)

    summary.to_csv(
        os.path.join(SAVE_DIR, f"{name}_quality_summary.csv"),
        encoding="utf-8-sig"
    )

    return summary


r1_summary = quality_summary(r1, "rater1")
r2_summary = quality_summary(r2, "rater2")

print("Rater1 品質摘要")
display(r1_summary)

print("Rater2 品質摘要")
display(r2_summary)


# =========================
# 8. 合併兩位評分者分數
# =========================
merge_cols = [ID_COL, "text", SOURCE_COL, "true_source_code"]

r1_renamed = r1[merge_cols + score_cols + ["overall_quality_score", "predicted_source_code"]].copy()
r2_renamed = r2[[ID_COL] + score_cols + ["overall_quality_score", "predicted_source_code"]].copy()

r1_renamed = r1_renamed.rename(columns={
    col: f"{col}_r1" for col in score_cols + ["overall_quality_score", "predicted_source_code"]
})

r2_renamed = r2_renamed.rename(columns={
    col: f"{col}_r2" for col in score_cols + ["overall_quality_score", "predicted_source_code"]
})

merged = r1_renamed.merge(r2_renamed, on=ID_COL, how="inner")

print("合併後筆數:", len(merged))



# =========================
# 9. 兩位評分者平均分數
# =========================
for col in score_cols + ["overall_quality_score"]:
    merged[f"{col}_avg"] = merged[[f"{col}_r1", f"{col}_r2"]].mean(axis=1)

avg_cols = [f"{col}_avg" for col in score_cols + ["overall_quality_score"]]

avg_summary = merged[avg_cols].agg(["mean", "std", "min", "max"]).T.round(3)

print("兩位評分者平均後品質摘要")
display(avg_summary)

avg_summary.to_csv(
    os.path.join(SAVE_DIR, "two_raters_average_quality_summary.csv"),
    encoding="utf-8-sig"
)


# =========================
# 10. 分 real / synthetic 比較平均品質
# =========================
source_avg_summary = merged.groupby(SOURCE_COL)[avg_cols].agg(["mean", "std", "count"]).round(3)

print("依 true_source 分組的平均品質")
display(source_avg_summary)

source_avg_summary.to_csv(
    os.path.join(SAVE_DIR, "two_raters_quality_by_source.csv"),
    encoding="utf-8-sig"
)


# =========================
# 11. 高品質比例
# overall_quality_score_avg >= 4
# =========================
HIGH_THRESHOLD = 4.0

high_quality_rate = (merged["overall_quality_score_avg"] >= HIGH_THRESHOLD).mean()
print(f"整體高品質比例 overall >= {HIGH_THRESHOLD}: {high_quality_rate:.2%}")

high_quality_by_source = merged.groupby(SOURCE_COL)["overall_quality_score_avg"].apply(
    lambda x: (x >= HIGH_THRESHOLD).mean()
).round(3)

print("分來源高品質比例：")
display(high_quality_by_source)


# =========================
# 12. Cohen's kappa：來源判斷一致性
# predicted_source: 0=real, 1=synthetic
# =========================
valid_kappa = merged.dropna(subset=["predicted_source_code_r1", "predicted_source_code_r2"])

kappa_source = cohen_kappa_score(
    valid_kappa["predicted_source_code_r1"],
    valid_kappa["predicted_source_code_r2"]
)

print(f"Cohen's kappa（predicted_source 來源判斷）: {kappa_source:.3f}")


# =========================
# 13. 1–5 分評分者一致性
# Spearman correlation + 加權 kappa
# =========================
agreement_rows = []

for col in score_cols:
    valid = merged.dropna(subset=[f"{col}_r1", f"{col}_r2"])

    spearman_corr, spearman_p = spearmanr(valid[f"{col}_r1"], valid[f"{col}_r2"])

    weighted_kappa = cohen_kappa_score(
        valid[f"{col}_r1"].astype(int),
        valid[f"{col}_r2"].astype(int),
        weights="quadratic"
    )

    agreement_rows.append({
        "score": col,
        "spearman_corr": spearman_corr,
        "spearman_p": spearman_p,
        "quadratic_weighted_kappa": weighted_kappa
    })

agreement_df = pd.DataFrame(agreement_rows).round(3)

print("1–5 分評分者一致性")
display(agreement_df)

agreement_df.to_csv(
    os.path.join(SAVE_DIR, "inter_rater_agreement_scores.csv"),
    index=False,
    encoding="utf-8-sig"
)


# =========================
# 14. 來源辨識率：兩位評分者各自的辨識率
# =========================
for name, df in [("rater1", r1), ("rater2", r2)]:
    valid = df.dropna(subset=["true_source_code", "predicted_source_code"])

    acc = accuracy_score(valid["true_source_code"], valid["predicted_source_code"])

    print(f"{name} 真實/合成辨識率: {acc:.2%}")

    cm = confusion_matrix(
        valid["true_source_code"],
        valid["predicted_source_code"],
        labels=[0, 1]
    )

    cm_df = pd.DataFrame(
        cm,
        index=["actual_real", "actual_synthetic"],
        columns=["pred_real", "pred_synthetic"]
    )

    display(cm_df)

    cm_df.to_csv(
        os.path.join(SAVE_DIR, f"{name}_source_confusion_matrix.csv"),
        encoding="utf-8-sig"
    )


# =========================
# 15. 合併結果輸出
# =========================
merged.to_excel(
    os.path.join(SAVE_DIR, "two_raters_merged_results.xlsx"),
    index=False
)

print("完成，所有結果已輸出到：", SAVE_DIR)
```

```python
# =========================
# 0. 匯入套件
# =========================
import os
import pandas as pd
import numpy as np
from sklearn.metrics import cohen_kappa_score, accuracy_score, confusion_matrix
from scipy.stats import spearmanr


# =========================
# 1. 檔案路徑
# =========================
RATER1_PATH = r"D:\時間序列\DECEN\human_quality_rating\human_quality_rating_blind_test_n.xlsx"
RATER2_PATH = r"D:\時間序列\DECEN\human_quality_rating\human_quality_rating_blind_test_w.xlsx"

# GPT 評分後的檔案
GPT_PATH = r"D:\時間序列\DECEN\human_quality_rating\human_quality_rating_blind_test_with_GPT_v1.xlsx"

SAVE_DIR = r"D:\時間序列\DECEN\human_quality_rating\results"
os.makedirs(SAVE_DIR, exist_ok=True)


# =========================
# 2. 讀取資料
# =========================
r1 = pd.read_excel(RATER1_PATH)
r2 = pd.read_excel(RATER2_PATH)
gpt = pd.read_excel(GPT_PATH)

print("Rater1 欄位:", r1.columns.tolist())
print("Rater2 欄位:", r2.columns.tolist())
print("GPT 欄位:", gpt.columns.tolist())


# =========================
# 3. 欄位設定
# =========================
ID_COL = "sample_id"
TEXT_COL = "text"
SOURCE_COL = "true_source"
PRED_COL = "predicted_source"

score_cols = [
    "fluency_score",
    "realism_score",
    "label_consistency_score"
]

gpt_score_cols = [
    "gpt_fluency_score",
    "gpt_realism_score",
    "gpt_label_consistency_score"
]

GPT_PRED_COL = "gpt_predicted_source"


# =========================
# 4. 檢查必要欄位
# =========================
for col in [ID_COL, SOURCE_COL, PRED_COL] + score_cols:
    if col not in r1.columns:
        raise ValueError(f"Rater1 缺少欄位：{col}")

for col in [ID_COL, PRED_COL] + score_cols:
    if col not in r2.columns:
        raise ValueError(f"Rater2 缺少欄位：{col}")

for col in [ID_COL] + gpt_score_cols + [GPT_PRED_COL]:
    if col not in gpt.columns:
        raise ValueError(f"GPT 檔案缺少欄位：{col}")


# =========================
# 5. rater2 / GPT 補回 true_source
# =========================
r2 = r2.merge(
    r1[[ID_COL, SOURCE_COL]],
    on=ID_COL,
    how="left"
)

gpt = gpt.merge(
    r1[[ID_COL, SOURCE_COL]],
    on=ID_COL,
    how="left"
)

print("Rater2 true_source：")
print(r2[SOURCE_COL].value_counts(dropna=False))

print("GPT true_source：")
print(gpt[SOURCE_COL].value_counts(dropna=False))


# =========================
# 6. 轉換來源欄位
# =========================
def normalize_source(x):
    if pd.isna(x):
        return np.nan

    x = str(x).strip().lower()

    if x in ["0", "real", "真實"]:
        return 0
    elif x in ["1", "synthetic", "合成"]:
        return 1
    else:
        return np.nan


# human rater
for df in [r1, r2]:
    for col in score_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    df["overall_quality_score"] = df[score_cols].mean(axis=1)
    df["true_source_code"] = df[SOURCE_COL].apply(normalize_source)
    df["predicted_source_code"] = df[PRED_COL].apply(normalize_source)


# GPT rater
for col in gpt_score_cols:
    gpt[col] = pd.to_numeric(gpt[col], errors="coerce")

gpt["gpt_overall_quality_score"] = gpt[gpt_score_cols].mean(axis=1)
gpt["true_source_code"] = gpt[SOURCE_COL].apply(normalize_source)
gpt["gpt_predicted_source_code"] = gpt[GPT_PRED_COL].apply(normalize_source)


# =========================
# 7. 各評分者品質摘要
# =========================
def quality_summary_human(df, name):
    summary = df[score_cols + ["overall_quality_score"]].agg(
        ["mean", "std", "min", "max"]
    ).T.round(3)

    summary.to_csv(
        os.path.join(SAVE_DIR, f"{name}_quality_summary.csv"),
        encoding="utf-8-sig"
    )

    return summary


def quality_summary_gpt(df):
    summary = df[gpt_score_cols + ["gpt_overall_quality_score"]].agg(
        ["mean", "std", "min", "max"]
    ).T.round(3)

    summary.to_csv(
        os.path.join(SAVE_DIR, "gpt_quality_summary.csv"),
        encoding="utf-8-sig"
    )

    return summary


print("Rater1 品質摘要")
display(quality_summary_human(r1, "rater1"))

print("Rater2 品質摘要")
display(quality_summary_human(r2, "rater2"))

print("GPT 品質摘要")
display(quality_summary_gpt(gpt))


# =========================
# 8. 合併三位評分者
# =========================
merge_cols = [ID_COL, TEXT_COL, SOURCE_COL, "true_source_code"]

r1_renamed = r1[
    merge_cols + score_cols + ["overall_quality_score", "predicted_source_code"]
].copy()

r2_renamed = r2[
    [ID_COL] + score_cols + ["overall_quality_score", "predicted_source_code"]
].copy()

gpt_renamed = gpt[
    [ID_COL] + gpt_score_cols + ["gpt_overall_quality_score", "gpt_predicted_source_code"]
].copy()


r1_renamed = r1_renamed.rename(columns={
    col: f"{col}_r1" for col in score_cols + ["overall_quality_score", "predicted_source_code"]
})

r2_renamed = r2_renamed.rename(columns={
    col: f"{col}_r2" for col in score_cols + ["overall_quality_score", "predicted_source_code"]
})

# GPT 欄位改成一致格式
gpt_renamed = gpt_renamed.rename(columns={
    "gpt_fluency_score": "fluency_score_gpt",
    "gpt_realism_score": "realism_score_gpt",
    "gpt_label_consistency_score": "label_consistency_score_gpt",
    "gpt_overall_quality_score": "overall_quality_score_gpt",
    "gpt_predicted_source_code": "predicted_source_code_gpt"
})

merged = r1_renamed.merge(r2_renamed, on=ID_COL, how="inner")
merged = merged.merge(gpt_renamed, on=ID_COL, how="inner")

print("合併後筆數:", len(merged))


# =========================
# 9. Human 平均分數
# =========================
for col in score_cols + ["overall_quality_score"]:
    merged[f"{col}_human_avg"] = merged[[f"{col}_r1", f"{col}_r2"]].mean(axis=1)

human_avg_cols = [f"{col}_human_avg" for col in score_cols + ["overall_quality_score"]]

human_avg_summary = merged[human_avg_cols].agg(["mean", "std", "min", "max"]).T.round(3)

print("兩位人工評分者平均品質摘要")
display(human_avg_summary)

human_avg_summary.to_csv(
    os.path.join(SAVE_DIR, "human_average_quality_summary.csv"),
    encoding="utf-8-sig"
)


# =========================
# 10. 三位評分者平均分數
# =========================
for col in score_cols + ["overall_quality_score"]:
    merged[f"{col}_all_avg"] = merged[
        [f"{col}_r1", f"{col}_r2", f"{col}_gpt"]
    ].mean(axis=1)

all_avg_cols = [f"{col}_all_avg" for col in score_cols + ["overall_quality_score"]]

all_avg_summary = merged[all_avg_cols].agg(["mean", "std", "min", "max"]).T.round(3)

print("Human1 + Human2 + GPT 三方平均品質摘要")
display(all_avg_summary)

all_avg_summary.to_csv(
    os.path.join(SAVE_DIR, "three_raters_average_quality_summary.csv"),
    encoding="utf-8-sig"
)


# =========================
# 11. 分 real / synthetic 比較
# =========================
source_human_avg_summary = merged.groupby(SOURCE_COL)[human_avg_cols].agg(
    ["mean", "std", "count"]
).round(3)

source_all_avg_summary = merged.groupby(SOURCE_COL)[all_avg_cols].agg(
    ["mean", "std", "count"]
).round(3)

print("依 true_source 分組：Human 平均")
display(source_human_avg_summary)

print("依 true_source 分組：三方平均")
display(source_all_avg_summary)

source_human_avg_summary.to_csv(
    os.path.join(SAVE_DIR, "human_quality_by_source.csv"),
    encoding="utf-8-sig"
)

source_all_avg_summary.to_csv(
    os.path.join(SAVE_DIR, "three_raters_quality_by_source.csv"),
    encoding="utf-8-sig"
)


# =========================
# 12. 高品質比例
# =========================
HIGH_THRESHOLD = 4.0

human_high_rate = (merged["overall_quality_score_human_avg"] >= HIGH_THRESHOLD).mean()
all_high_rate = (merged["overall_quality_score_all_avg"] >= HIGH_THRESHOLD).mean()

print(f"Human 平均高品質比例 overall >= {HIGH_THRESHOLD}: {human_high_rate:.2%}")
print(f"三方平均高品質比例 overall >= {HIGH_THRESHOLD}: {all_high_rate:.2%}")

human_high_by_source = merged.groupby(SOURCE_COL)["overall_quality_score_human_avg"].apply(
    lambda x: (x >= HIGH_THRESHOLD).mean()
).round(3)

all_high_by_source = merged.groupby(SOURCE_COL)["overall_quality_score_all_avg"].apply(
    lambda x: (x >= HIGH_THRESHOLD).mean()
).round(3)

print("Human 分來源高品質比例：")
display(human_high_by_source)

print("三方分來源高品質比例：")
display(all_high_by_source)


# =========================
# 13. predicted_source 一致性：Cohen's kappa
# =========================
pairs = [
    ("rater1", "rater2", "predicted_source_code_r1", "predicted_source_code_r2"),
    ("rater1", "gpt", "predicted_source_code_r1", "predicted_source_code_gpt"),
    ("rater2", "gpt", "predicted_source_code_r2", "predicted_source_code_gpt"),
]

source_kappa_rows = []

for a, b, col_a, col_b in pairs:
    valid = merged.dropna(subset=[col_a, col_b])

    kappa = cohen_kappa_score(valid[col_a], valid[col_b])

    raw_agreement = (valid[col_a] == valid[col_b]).mean()

    source_kappa_rows.append({
        "pair": f"{a}_vs_{b}",
        "cohen_kappa": round(kappa, 3),
        "raw_agreement": round(raw_agreement, 3)
    })

source_kappa_df = pd.DataFrame(source_kappa_rows)

print("predicted_source 來源判斷一致性")
display(source_kappa_df)

source_kappa_df.to_csv(
    os.path.join(SAVE_DIR, "source_prediction_kappa_all_pairs.csv"),
    index=False,
    encoding="utf-8-sig"
)


# =========================
# 14. 1–5分一致性：Spearman + Quadratic Weighted Kappa
# =========================
agreement_rows = []

score_pairs = [
    ("rater1", "rater2", "_r1", "_r2"),
    ("rater1", "gpt", "_r1", "_gpt"),
    ("rater2", "gpt", "_r2", "_gpt"),
]

for score in score_cols:
    for a, b, suffix_a, suffix_b in score_pairs:
        col_a = f"{score}{suffix_a}"
        col_b = f"{score}{suffix_b}"

        valid = merged.dropna(subset=[col_a, col_b]).copy()

        spearman_corr, spearman_p = spearmanr(valid[col_a], valid[col_b])

        weighted_kappa = cohen_kappa_score(
            valid[col_a].astype(int),
            valid[col_b].astype(int),
            weights="quadratic"
        )

        agreement_rows.append({
            "score": score,
            "pair": f"{a}_vs_{b}",
            "spearman_corr": round(spearman_corr, 3),
            "spearman_p": round(spearman_p, 3),
            "quadratic_weighted_kappa": round(weighted_kappa, 3)
        })

agreement_df = pd.DataFrame(agreement_rows)

print("1–5 分評分者一致性")
display(agreement_df)

agreement_df.to_csv(
    os.path.join(SAVE_DIR, "inter_rater_agreement_with_gpt.csv"),
    index=False,
    encoding="utf-8-sig"
)


# =========================
# 15. 來源辨識率 accuracy
# =========================
evaluators = [
    ("rater1", r1, "predicted_source_code"),
    ("rater2", r2, "predicted_source_code"),
    ("gpt", gpt, "gpt_predicted_source_code")
]

for name, df_eval, pred_code_col in evaluators:
    valid = df_eval.dropna(subset=["true_source_code", pred_code_col])

    acc = accuracy_score(valid["true_source_code"], valid[pred_code_col])

    print(f"{name} 真實/合成辨識率: {acc:.2%}")

    cm = confusion_matrix(
        valid["true_source_code"],
        valid[pred_code_col],
        labels=[0, 1]
    )

    cm_df = pd.DataFrame(
        cm,
        index=["actual_real", "actual_synthetic"],
        columns=["pred_real", "pred_synthetic"]
    )

    display(cm_df)

    cm_df.to_csv(
        os.path.join(SAVE_DIR, f"{name}_source_confusion_matrix.csv"),
        encoding="utf-8-sig"
    )


# =========================
# 16. Human 平均 vs GPT：Spearman
# =========================
human_gpt_corr_rows = []

for col in score_cols + ["overall_quality_score"]:
    human_col = f"{col}_human_avg"
    gpt_col = f"{col}_gpt"

    valid = merged.dropna(subset=[human_col, gpt_col])

    corr, p = spearmanr(valid[human_col], valid[gpt_col])

    human_gpt_corr_rows.append({
        "score": col,
        "human_avg_vs_gpt_spearman": round(corr, 3),
        "p_value": round(p, 3)
    })

human_gpt_corr_df = pd.DataFrame(human_gpt_corr_rows)

print("Human 平均 vs GPT 相關")
display(human_gpt_corr_df)

human_gpt_corr_df.to_csv(
    os.path.join(SAVE_DIR, "human_avg_vs_gpt_spearman.csv"),
    index=False,
    encoding="utf-8-sig"
)


# =========================
# 17. 輸出總表
# =========================
merged.to_excel(
    os.path.join(SAVE_DIR, "three_raters_merged_results.xlsx"),
    index=False
)

print("完成，所有結果已輸出到：", SAVE_DIR)
```

```python
# ==========================================
# 0. 安裝套件
# ==========================================
# !pip install openai pandas openpyxl tqdm
# ==========================================
# 1. import
# ==========================================

import pandas as pd
import json
import time

from tqdm import tqdm
from openai import OpenAI


# ==========================================
# 2. API KEY
# ==========================================

client = OpenAI(
    api_key="REPLACE_WITH_OPENAI_API_KEY"
)

# ==========================================
# 3. 讀取資料
# ==========================================

df = pd.read_excel(
    "D:\時間序列\DECEN\human_quality_rating\human_quality_rating_blind_test.xlsx"
)

TEXT_COL = "text"
print("資料數：",len(df))
# display(df.head(3))

# ==========================================
# 4. SYSTEM PROMPT
# ==========================================

SYSTEM_PROMPT = """
你是一位中文社群媒體心理健康研究評分者。

任務：
請依據操作型定義，
對中文 Instagram 憂鬱相關貼文進行品質評估。

注意：
單一特徵不得直接判斷。
請充分使用完整評分尺度（1~5），避免集中使用3或4分。
若貼文非常符合或非常不符合定義，請大膽使用1或5分。

僅評估：
1 fluency_score
2 realism_score
3 label_consistency_score
4 predicted_source

以下為具體分數依據：
1、fluency_score 流暢度
評估文本在中文語法與社群網路口語習慣下是否通順。
注意：憂鬱貼文本來可能有破碎感，不要因為情緒低落或思緒跳躍就扣分；只針對語法錯誤、AI 翻譯腔、不自然表達扣分。

1分：語句混亂、主謂賓脫節、邏輯矛盾，難以理解。
2分：有明顯語病、簡轉繁殘留、兩岸用語混雜，或機械式重複句。
3分：可理解但生硬，像心理諮商個案報告、臨床診斷書或教科書式陳述。
4分：大致自然通順，符合現代網路口語，標點與斷句自然。
5分：非常自然，像台灣本地網路使用者隨筆，含自然語氣詞、口語贅字或情緒化表達，無結構化痕跡。

2、realism_score 真實感
評估文本是否像真實存在的人在 Instagram 上的日常分享，而非大量生成的模板句。

1分：明顯像 AI 或模板句，出現求助建議、免責聲明、JSON 格式、罐頭訊息或過度對稱排版。
2分：只有情緒字眼堆砌，缺乏生活事件背景或導因。
3分：有部分真實感，但事件背景籠統模糊，例如「今天很糟」「遇到很多煩心事」。
4分：大致像真實社群貼文，有具體心理社會情境，如學業壓力、人際衝突、家庭爭吵、失眠、工作壓力等，並有個人行為描述。
5分：非常像真實 IG/社群貼文，具有本地生活細節、平台生態特徵、自然斷行、長短句交錯或極具個人感的情緒表達。

3、label_consistency_score 憂鬱一致性
評估文本是否對齊 DSM-5 相關憂鬱核心情緒向度：
D = 情緒低落
A = 快樂缺失
W = 無價值感或罪惡感
U = 自殺意念

1分：幾乎無憂鬱語意，屬於正面、中性日常，或單一事件的短暫抱怨。
2分：只有一般負面情緒，如累、孤單、無聊、生氣，但無法明確對應 D/A/W/U。
3分：有部分憂鬱線索，例如提不起勁、微弱自責、隱約失去興趣，但力道較弱。
4分：明確符合至少一項核心向度，例如長期絕望空虛、明顯快樂缺失、強烈自我貶低、含蓄想消失或不想醒來。
5分：明確且多向度呈現憂鬱語意，情緒與生活事件高度融合，例如人際衝突引發無價值感，同時伴隨失眠、絕望或自殺意念。

4、predicted_source 來源判斷
請進行盲測判斷。
請特別注意：本任務中的合成資料通常不是語法錯誤，而是「太完整、太平順、太像在描述憂鬱症狀」。
在 predicted_source 判斷上，請避免過度保守。
若文本同時具有自然語法與明顯症狀模板感，請判為 synthetic。

請優先將下列情況判為合成資料 1：
- 結構過於完整、對稱、起承轉合太清楚
- 連續使用相同句型或工整排比
- 語句雖自然，但像整理過的心理狀態描述
- 直接「點出」情緒名詞，或是情緒詞連發，例如：無力感、罪惡感、價值感、情緒低落、缺乏動力、想消失、想結束一切
- 多個句子結構相似，語氣穩定，缺少真人即時發文的雜亂感
- 貼文結尾過於總結式、理性化或完整

將下列情況判為真實資料 0：
- 有明顯生活碎片或平台語境
- 有自然口語、髒話、標點堆疊或混亂斷句
- 有突兀轉折、跳躍思緒、碎碎念、句子長短不一
- 情緒表達不工整，可能只是在某個片刻發洩
- 有具體台灣在地生活細節、口語語助詞、錯字、標點混亂、斷句不規則

注意：
- 不要因為文句通順就判為真實。
- 不要因為有生活事件就一定判為真實。
- 若文本「自然但過度完整地呈現憂鬱症狀」，應偏向判為合成資料。
- 若無法確定，請根據是否具有「模板化的症狀描述」判斷


請先內部完成：
1 是否具有生活事件背景
2 是否符合 DSM5 D/A/W/U
3 是否符合自然中文社群表達
4 是否具有模板化特徵

不要輸出推理。
只輸出 JSON：

{
"fluency_score":1~5,
"realism_score":1~5,
"label_consistency_score":1~5,
"predicted_source":0或1
}

"""

# ==========================================
# 5. GPT評分函數
# ==========================================

def evaluate_post(text):

    user_prompt=f"""

請評估以下Instagram貼文：

{text}

"""

    for retry in range(3):
        try:
            response=client.chat.completions.create(
                model="gpt-4.1",
                temperature=0,
                response_format={
                    "type":"json_object"
                },
                messages=[
                    {
                        "role":"system",
                        "content":SYSTEM_PROMPT
                    },
                    {
                        "role":"user",
                        "content":user_prompt
                    }
                ]
            )

            result=response.choices[0].message.content
            result=json.loads(result)
            return result

        except Exception as e:
            print("Retry:",retry,e)
            time.sleep(3)

    return {
        "fluency_score":None,
        "realism_score":None,
        "label_consistency_score":None,
        "predicted_source":None
    }


# ==========================================
# 6. 批次評分
# ==========================================
# results=[]

# for text in tqdm(df[TEXT_COL]):
#     score=evaluate_post(text)
#     results.append(score)
#     time.sleep(0.5)

# ==========================================
# 試跑前5筆
# ==========================================
TEST_N=100
test_df=df.head(TEST_N).copy()
results=[]

for i,text in enumerate(
    tqdm(test_df[TEXT_COL])
):

    print()
    print("="*50)
    print("Sample",i+1)
    print("="*50)
    score=evaluate_post(text)

    print("原文：")
    print(text[:300])
    print()
    print("GPT評分：")
    print(score)
    results.append(score)
    time.sleep(0.5)


gpt_df=pd.DataFrame(results)

gpt_df.columns=[
    "gpt_fluency_score",
    "gpt_realism_score",
    "gpt_label_consistency_score",
    "gpt_predicted_source"
]


test_result=pd.concat(
    [
        test_df.reset_index(drop=True),
        gpt_df
    ],
    axis=1
)
display(test_result)

gpt_df=pd.DataFrame(results)

# ==========================================
# 7. 合併結果
# ==========================================

gpt_df.columns=[
    "gpt_fluency_score",
    "gpt_realism_score",
    "gpt_label_consistency_score",
    "gpt_predicted_source"
]


final_df=pd.concat(
    [
        df,
        gpt_df
    ],
    axis=1

)

# ==========================================
# 8. 存檔
# ==========================================
OUTPUT_PATH="D:\時間序列\DECEN\human_quality_rating\human_quality_rating_blind_test_with_GPT.xlsx"

final_df.to_excel(
    OUTPUT_PATH,
    index=False

)

print()
print("完成")
print(OUTPUT_PATH)
display(final_df.head())
```
